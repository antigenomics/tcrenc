```python
import torch
```

```python
path_in = '../tcrenc/models/autoencoder_kidera/weights/epitope.pth'
w_dict = torch.load(path_in, map_location=torch.device('cpu'))
```

```python
for key in list(w_dict.keys()):
    if 'encoder' in key:
        w_dict[key.replace('encoder.', 'encoder.encoder.')] = w_dict.pop(key)
    if 'linear_encode' in key:
        w_dict[key.replace('linear_encode.', 'encoder.linear_encode.')] = w_dict.pop(key)
    if 'decoder' in key:
        w_dict[key.replace('decoder.', 'decoder.decoder.')] = w_dict.pop(key)
    if 'linear_decode' in key:
        w_dict[key.replace('linear_decode.', 'decoder.linear_decode.')] = w_dict.pop(key)
```

```python
path_out = '../tcrenc/models/autoencoder_kidera/weights/epitope_restructed.pth'
torch.save(w_dict, path_out)
```
